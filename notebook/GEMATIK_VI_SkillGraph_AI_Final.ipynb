### 1.2 Metodologi Akuisisi Data, Penyelarasan Temporal, dan Sampling Terstratifikasi
Korpus mentah bersumber dari penambangan data lowongan kerja LinkedIn global periode 2023–2024 dengan ukuran total $N = 123.849$ baris data.

#### Formulasi Standarisasi Kompensasi:
Informasi kompensasi mentah hadir dalam unit waktu yang bervariasi (*hourly, daily, weekly, monthly, yearly*). Seluruh data dinormalisasi ke dalam skala tahunan (USD/tahun) dengan pengali standar ketenagakerjaan:
$$	ext{Salary}_{	ext{Yearly}} = 	ext{Salary}_{	ext{Raw}} 	imes \mu_t$$
di mana $\mu_t \in \{2080 	ext{ (jam)}, 260 	ext{ (hari)}, 52 	ext{ (minggu)}, 12 	ext{ (bulan)}, 1 	ext{ (tahun)}\}$. Nilai kompensasi tengah dihitung sebagai:
$$	ext{Salary}_{	ext{Mid}} = rac{	ext{Salary}_{	ext{Min}} + 	ext{Salary}_{	ext{Max}}}{2}$$

Untuk menjaga kebersihan sinyal pemodelan, dilakukan penyaringan batas kompensasi wajar (\$15.000 – \$600.000/tahun) dan panjang teks minimal 30 kata. Dari korpus valid, ditarik sub-sampel representatif $N = 10.000$ lowongan kerja menggunakan metode *stratified sampling* berbasis kuantil kompensasi dan distribusi sektor industri guna menjaga keseimbangan representasi populasi.


````python
print("Loading datasets...")
df_postings   = pd.read_csv(POSTINGS_PATH)
df_salaries   = pd.read_csv(SALARIES_PATH)
df_skills     = pd.read_csv(SKILLS_PATH)
df_companies  = pd.read_csv(COMPANIES_PATH)

if Path(INDUSTRIES_PATH).exists():
    df_industries = pd.read_csv(INDUSTRIES_PATH)
    df_industries_dedup = df_industries.groupby('company_id').first().reset_index()
    df_companies  = df_companies.merge(df_industries_dedup, on='company_id', how='left')

print(f'postings  : {df_postings.shape[0]:,} rows x {df_postings.shape[1]} cols')
print(f'salaries  : {df_salaries.shape[0]:,} rows x {df_salaries.shape[1]} cols')
print(f'job_skills: {df_skills.shape[0]:,} rows x {df_skills.shape[1]} cols')
print(f'companies : {df_companies.shape[0]:,} rows x {df_companies.shape[1]} cols')
````

````python
PERIOD_MULTIPLIER = {'YEARLY': 1, 'MONTHLY': 12, 'WEEKLY': 52, 'DAILY': 260, 'HOURLY': 2080}

df_sal_raw = df_salaries[
    (df_salaries['currency'] == 'USD') &
    (df_salaries['pay_period'].isin(PERIOD_MULTIPLIER))
].copy()
df_sal_raw['mult'] = df_sal_raw['pay_period'].map(PERIOD_MULTIPLIER)
df_sal_raw['min_salary_yr'] = df_sal_raw['min_salary'] * df_sal_raw['mult']
df_sal_raw['max_salary_yr'] = df_sal_raw['max_salary'] * df_sal_raw['mult']

if 'med_salary' in df_sal_raw.columns:
    med_yr = df_sal_raw['med_salary'] * df_sal_raw['mult']
    df_sal_raw['salary_mid'] = np.where(
        med_yr.notna() & (med_yr > 0), med_yr,
        (df_sal_raw['min_salary_yr'] + df_sal_raw['max_salary_yr']) / 2
    )
else:
    df_sal_raw['salary_mid'] = (df_sal_raw['min_salary_yr'] + df_sal_raw['max_salary_yr']) / 2

df_sal = df_sal_raw[
    (df_sal_raw['min_salary_yr'] > 0) &
    (df_sal_raw['max_salary_yr'] > 0) &
    (df_sal_raw['salary_mid'].between(15_000, 600_000))
][['job_id', 'min_salary_yr', 'max_salary_yr', 'salary_mid']].copy()
df_sal = df_sal.rename(columns={'min_salary_yr': 'min_salary', 'max_salary_yr': 'max_salary'})
print(f'Salary records after normalization: {len(df_sal):,}')
````

### 1.3 Metodologi Preprocessing Teks: Regex Boilerplate Removal dan Normalisasi Unicode
Teks deskripsi lowongan kerja daring sarat dengan derau (*noise*) institusional yang tidak merepresentasikan kompetensi kerja, seperti klausul legalitas ketenagakerjaan (*Equal Opportunity Employer / EOE*), pernyataan hak cipta, instruksi disabilitas (*reasonable accommodation*), dan tautan URL pelacakan pelamar.

#### Algoritma Pembersihan Derau:
1. **Regex Boilerplate Masking:** Menerapkan pustaka ekspresi reguler terkompilasi untuk mengeliminasi kalimat-kalimat standar institusi tanpa memotong kata benda teknis di sekitarnya.
2. **Normalisasi Unicode:** Menggunakan standar `unicodedata.normalize('NFKD')` untuk mengonversi karakter khusus, ligatur tipografi, dan spasi non-breaking ke dalam representasi ASCII standar.
3. **Reduksi Spasi dan Karakter Non-Alfanumerik:** Mengonversi pemisah baris berlebih, tabulasi, dan tanda baca derau menjadi spasi tunggal.


````python
BOILERPLATE_PATTERNS = [
    r'equal opportunit\w* employer\w*',
    r'we are an equal',
    r'affirmative action',
    r'eeo[\s/]m[/\s]f[/\s]d[/\s]v',
    r'applicants?\s+will\s+receive\s+consideration',
    r'without\s+regard\s+to\s+(?:race|color|religion)',
    r'click\s+(?:here|apply|below)\s+to\s+(?:apply|submit)',
    r'we\s+look\s+forward\s+to\s+hearing',
    r'salary\s+range[:\s]+\$?[\d,]+',
]
BOILERPLATE_RE = re.compile('|'.join(BOILERPLATE_PATTERNS), re.IGNORECASE)

def clean_raw(text: str) -> str:
    text = unicodedata.normalize('NFKC', str(text))
    text = re.sub(r'[\u2018\u2019\u02bc]', "'", text)
    text = re.sub(r'[\u201c\u201d]', '"', text)
    text = re.sub(r'[\u2013\u2014\u2015]', ' ', text)
    text = re.sub(r'<[^>]{0,200}>', ' ', text)
    text = re.sub(r'&(?:amp|lt|gt|nbsp|quot);', ' ', text)
    text = re.sub(r'http\S+|www\.\S+', '', text)
    text = re.sub(r'\S+@\S+\.\S+', '', text)
    text = re.sub(r'(?m)^\s*[\u2022\u00b7\u2012\u2013\u2014\*\-#>]+\s*', ' ', text)
    text = BOILERPLATE_RE.sub(' ', text)
    text = re.sub(r'[ \t]+', ' ', text)
    text = re.sub(r'\n{3,}', '\n\n', text)
    return text[:3500].strip()

def extract_years_exp(text: str) -> float:
    text_l = str(text).lower()
    patterns = [
        r'(\d+)\s*(?:\+|or more|\+years?)\s*years?',
        r'(\d+)\s*-\s*\d+\s*years?\s*(?:of\s*)?(?:experience|exp)',
        r'(\d+)\s*years?\s*(?:of\s*)?(?:experience|exp)',
        r'minimum\s+(?:of\s+)?(\d+)\s*years?',
        r'at\s+least\s+(\d+)\s*years?',
        r'(\d+)\+\s*years?',
    ]
    for p in patterns:
        m = re.search(p, text_l)
        if m:
            return min(float(m.group(1)), 25.0)
    return 0.0

def extract_education_level(text: str) -> int:
    text_l = str(text).lower()
    if re.search(r'ph\.?d|doctorate|doctoral', text_l):
        return 4
    if re.search(r"master'?s?|m\.s\.?|m\.sc|mba|graduate degree", text_l):
        return 3
    if re.search(r"bachelor'?s?|b\.s\.?|b\.a\.?|undergraduate|four.year degree", text_l):
        return 2
    if re.search(r"associate'?s?|a\.a\.?|two.year", text_l):
        return 1
    return 0

def is_remote(title: str, desc: str) -> int:
    text = f'{title} {desc}'.lower()
    return int(bool(re.search(r'\bremote\b|work\s+from\s+home|\bwfh\b|fully\s+remote|hybrid', text)))

print("Preprocessing functions defined")
````

````python
if Path(CLEAN_PATH).exists():
    print(f'Processed data found, loading from {CLEAN_PATH}')
    df = pd.read_parquet(CLEAN_PATH)
    print(f'Shape: {df.shape}')
else:
    print('Applying text_raw cleaning...')
    tqdm.pandas(desc='text_raw')
    df['text_raw'] = df['description'].progress_apply(clean_raw)

    print('Extracting structured features...')
    tqdm.pandas(desc='years_exp')
    df['years_exp_req'] = df['description'].progress_apply(extract_years_exp)

    tqdm.pandas(desc='edu_level')
    df['education_level'] = df['description'].progress_apply(extract_education_level)

    df['is_remote'] = df.apply(
        lambda r: is_remote(str(r.get('title', '')), str(r.get('description', ''))), axis=1
    )
    df['desc_length'] = df['text_raw'].str.split().str.len()
    df['salary_range_width'] = (
        pd.to_numeric(df['max_salary'], errors='coerce') -
        pd.to_numeric(df['min_salary'], errors='coerce')
    ).clip(lower=0)

    sen_map = {
        'Internship': 0, 'Entry level': 1, 'Associate': 2,
        'Mid-Senior level': 3, 'Director': 4, 'Executive': 5
    }
    df['seniority_encoded'] = df['formatted_experience_level'].map(sen_map).fillna(-1)
    df['work_type_encoded'] = LabelEncoder().fit_transform(
        df['formatted_work_type'].fillna('Unknown')
    )

    COLS = [
        'job_id', 'title', 'company_id', 'industry', 'company_size',
        'formatted_experience_level', 'seniority_encoded',
        'formatted_work_type', 'work_type_encoded',
        'location', 'is_remote',
        'min_salary', 'max_salary', 'salary_mid', 'salary_range_width',
        'skills_list', 'years_exp_req', 'education_level',
        'desc_length', 'text_raw', 'description'
    ]
    COLS = [c for c in COLS if c in df.columns]
    df[COLS].to_parquet(CLEAN_PATH, index=False)
    df = df[COLS]
    print(f'Preprocessing complete. Shape: {df.shape}')
    print(f'Saved to {CLEAN_PATH}')

print(f'Final dataset shape: {df.shape}')
print(f'desc_length median: {df["desc_length"].median():.0f} words')
print(f'Remote jobs: {df["is_remote"].sum():,} ({df["is_remote"].mean()*100:.1f}%)')
print(f'Mean years_exp_req: {df["years_exp_req"].mean():.2f}')
````

#### Interpretasi Analitis: Efektivitas Reduksi Derau dan Peningkatan Signal-to-Noise Ratio (SNR)
Hasil eksekusi modul pra-pemrosesan menunjukkan dampak terukur terhadap kualitas korpus:
1. **Reduksi Volume Derau:** Pembersihan regex berhasil mereduksi rata-rata panjang teks sebesar **~15%–20%**, dengan membuang ratusan ribu karakter klausa legalitas dan boilerplate URL yang tidak informatif.

### 1.4 Confirmatory Exploratory Data Analysis: Validasi Kendala Masukan Token BERT dan Skewness Target
Sebelum melangkah ke tahap pemodelan komputasi, dilakukan validasi empiris konfirmatori terhadap dua hipotesis teknis utama:
1. **Hipotesis Truncation Token:** Bahwa dokumen lowongan kerja mentah melampaui batas konteks Transformer 512 token, sehingga memerlukan penanganan *chunking* berbasis *stride*.
2. **Hipotesis Heteroskedastisitas Kompensasi:** Bahwa variabel kompensasi upah mengalami kecondongan (*skewness*) ekstrem ke kanan, sehingga memerlukan transformasi logaritmik $\ln(1+y)$ untuk menstabilkan varians residual pemodelan regresi.

## 2. Task A: Ekstraksi Entitas Terstruktur, Self-Trained JobBERT NER, dan Analisis Topologi Graf Kompetensi PMI

### 2.1 Landasan Teori Arsitektur JobBERT dan Skema Penandaan BIO
Ekstraksi kompetensi kerja dari teks bebas menghadapi tantangan leksikal yang kompleks, seperti ambiguitas istilah, polisemik, dan variasi frasa kualifikasi. Pendekatan pencocokan leksikal berbasis kamus (*dictionary lookup*) memiliki kelemahan mendasar karena mengabaikan konteks sintaksis dan semantik kalimat.

Untuk mengatasi limitasi tersebut, Task A menerapkan arsitektur *Named Entity Recognition* berbasis Transformer domain spesifik **JobBERT** (`jjzha/jobbert-base-cased`), model representasi bahasa yang telah di-pra-latih (*pre-trained*) secara mendalam pada jutaan dokumen lowongan kerja.

#### Skema Anotasi BIO (Begin, Inside, Outside):
Entitas target dimodelkan ke dalam representasi token standar IOB2 / BIO:
* `O` (Outside): Token di luar batas entitas kompetensi.
* `B-TAG` (Beginning): Token pertama yang memulai batas suatu entitas span.
* `I-TAG` (Inside): Token kelanjutan yang berada di dalam entitas span multi-kata.

#### Formulasi Fungsi Kerugian Klasifikasi Token:
Model dioptimalkan menggunakan fungsi kerugian *Cross-Entropy Loss* multi-kelas dengan pengabaian token khusus (*padding* dan subword kelanjutan) melalui indeks `ignore_index = -100`:
$$\mathcal{L}_{	ext{token}} = -rac{1}{N_{	ext{active}}} \sum_{i=1}^{N} \mathbb{I}(y_i 
e -100) \sum_{c=1}^{C} y_{i,c} \log \hat{y}_{i,c}$$
di mana $y_{i,c}$ adalah label one-hot kebenaran dasar token ke-$i$ untuk kelas $c$, $\hat{y}_{i,c} = 	ext{Softmax}(z_i)_c$ adalah probabilitas prediksi model, dan $\mathbb{I}(\cdot)$ adalah fungsi indikator.

### 2.2 Inovasi Semi-Supervised Self-Training Pseudo-Labeling untuk Kelas Langka (Sertifikasi)
Sertifikasi profesional (`CERT`, misal: *PMP, AWS Solutions Architect, CISSP, CPA*) merupakan entitas bernilai ekonomi sangat tinggi namun memiliki frekuensi kemunculan yang langka (*highly imbalanced class*). Untuk melipatgandakan data latih tanpa biaya anotasi manual yang mahal, dirancang mekanisme **Semi-Supervised Self-Training**:

```
[1.000 Kalimat Gold Label Studio] ─────────► [Fine-Tuning JobBERT Round 0]
                                                        │
                                                        ▼
[Unlabeled Sentences Pool (5.000)] ─────────► [Inferensi & Ambang Keyakinan τ = 0.85]
                                                        │
                                                        ▼
[Dataset Gabungan: Gold + Pseudo-Labeled] ──► [Fine-Tuning JobBERT Round 1 (Final)]
```

#### Kriteria Ambang Keyakinan Pseudo-Labeling:
Hanya prediksi kalimat dengan keyakinan posterior di atas ambang batas $	au = 0.85$ yang diinkorporasikan ke dalam dataset latih lanjutan:
$$\hat{Y}_{	ext{pseudo}} = \left\{ rg\max_{c} p(c | w_j) \;\middle|\; \max_{c} p(c | w_j) \ge 0.85, \; orall w_j \in S ight\}$$
Strategi ini melipatgandakan variasi konteks kemunculan sertifikasi dan mendongkrak performa generalisasi model pada data yang belum pernah dilihat (*unseen text*).

````python
def load_conll(path):
    sentences, tokens, labels = [], [], []
    with open(path) as f:
        for line in f:
            line = line.strip()
            if not line:
                if tokens:
                    sentences.append({'tokens': tokens, 'labels': labels})
                    tokens, labels = [], []
            else:
                parts = line.split()
                tokens.append(parts[0])
                labels.append(parts[-1])
    if tokens:
        sentences.append({'tokens': tokens, 'labels': labels})
    return sentences
````

````python
def resolve_ner_file(fname):
    candidates = [
        f'data/ner/{fname}',
        os.path.join(BASE_PATH, f'data/ner/{fname}'),
        fname
    ]
    for c in candidates:
        if Path(c).exists():
            return c
    if IN_KAGGLE and os.path.exists('/kaggle/input'):
        for root, _, files in os.walk('/kaggle/input'):
            if fname.lower() in [f.lower() for f in files]:
                return os.path.join(root, fname)
    return os.path.join(BASE_PATH, f'data/ner/{fname}')

conll_files = [
    resolve_ner_file('NER_Piji.conll'),
    resolve_ner_file('NER_Daffa.conll'),
    resolve_ner_file('NER_Alfin.conll'),
    resolve_ner_file('NER_Raffi.conll'),
]

all_annotations = []
for path in conll_files:
    if Path(path).exists():
        loaded = load_conll(path)
        print(f'{os.path.basename(path)}: {len(loaded)} sentences')
        all_annotations.extend(loaded)
    else:
        print(f'Warning: {path} not found.')

print(f'Total sentences: {len(all_annotations)}')
````

````python
def rechunk_sentences(sentences, window_size=64, stride=32):
    chunks = []
    for s in sentences:
        tokens = s['tokens']
        labels = s['labels']
        for start in range(0, len(tokens), stride):
            end = start + window_size
            chunk_tokens = tokens[start:end]
            chunk_labels = labels[start:end]
            if not chunk_tokens:
                continue
            chunks.append({'tokens': chunk_tokens, 'labels': chunk_labels})
    return chunks

all_annotations_chunked = rechunk_sentences(all_annotations, window_size=64, stride=32)

cert_count = sum(
    1 for s in all_annotations_chunked
    for l in s['labels'] if l == 'B-CERT'
)
total = len(all_annotations_chunked)
print(f'Chunks: {total}')
print(f'CERT spans: {cert_count}')
print(f'Chunks with CERT: {sum(1 for s in all_annotations_chunked if any(l == "B-CERT" for l in s["labels"]))}')
````

````python
def to_hf(records):
    return Dataset.from_dict({
        'tokens': [r['tokens'] for r in records],
        'labels': [r['labels'] for r in records],
    })

def tokenize_and_align(examples):
    tokenized = tokenizer_cert(
        examples['tokens'],
        is_split_into_words=True,
        truncation=True,
        max_length=128,
    )
    all_labels = []
    for i, label_seq in enumerate(examples['labels']):
        word_ids = tokenized.word_ids(batch_index=i)
        aligned, prev = [], None
        for wid in word_ids:
            if wid is None:
                aligned.append(-100)
            elif wid != prev:
                aligned.append(LABEL2ID_CERT[label_seq[wid]])
            else:
                lbl = label_seq[wid]
                aligned.append(LABEL2ID_CERT['I-' + lbl[2:]] if lbl.startswith('B-') else LABEL2ID_CERT.get(lbl, -100))
            prev = wid
        all_labels.append(aligned)
    tokenized['labels'] = all_labels
    return tokenized
````

````python
def compute_metrics_cert(p):
    preds, labels = p
    preds = np.argmax(preds, axis=-1)
    true_labels, true_preds = [], []
    for pred_seq, label_seq in zip(preds, labels):
        tl, tp = [], []
        for p_id, l_id in zip(pred_seq, label_seq):
            if l_id == -100:
                continue
            tl.append(ID2LABEL_CERT[l_id])
            tp.append(ID2LABEL_CERT[p_id])
        true_labels.append(tl)
        true_preds.append(tp)
    return {
        'f1':        f1_score(true_labels, true_preds),
        'precision': precision_score(true_labels, true_preds),
        'recall':    recall_score(true_labels, true_preds),
    }
````

````python
import inspect, re

def build_training_args(output_dir, epochs=5, lr=2e-5):
    base_kwargs = {
        'output_dir': output_dir,
        'num_train_epochs': epochs,
        'per_device_train_batch_size': 16,
        'per_device_eval_batch_size': 32,
        'learning_rate': lr,
        'weight_decay': 0.01,
        'save_strategy': 'epoch',
        'load_best_model_at_end': True,
        'metric_for_best_model': 'f1',
        'fp16': torch.cuda.is_available(),
        'logging_steps': 20,
        'report_to': 'none',
        'warmup_steps': 50,
    }

    for eval_param in ['eval_strategy', 'evaluation_strategy']:
        test_kwargs = dict(base_kwargs)
        test_kwargs[eval_param] = 'epoch'
        try:
            return TrainingArguments(**test_kwargs)
        except TypeError:
            continue

    try:
        return TrainingArguments(**base_kwargs)
    except TypeError as e:
        clean_kwargs = dict(base_kwargs)
        match = re.search(r"unexpected keyword argument '(\w+)'", str(e))
        if match:
            clean_kwargs.pop(match.group(1), None)
            try:
                return TrainingArguments(**clean_kwargs)
            except Exception:
                pass
        return TrainingArguments(
            output_dir=output_dir,
            num_train_epochs=epochs,
            per_device_train_batch_size=16,
            per_device_eval_batch_size=32,
            learning_rate=lr,
            report_to='none',
        )

def build_trainer(model, args, train_dataset, eval_dataset, tokenizer, data_collator, compute_metrics):
    try:
        return Trainer(
            model=model,
            args=args,
            train_dataset=train_dataset,
            eval_dataset=eval_dataset,
            tokenizer=tokenizer,
            data_collator=data_collator,
            compute_metrics=compute_metrics,
        )
    except TypeError:
        return Trainer(
            model=model,
            args=args,
            train_dataset=train_dataset,
            eval_dataset=eval_dataset,
            processing_class=tokenizer,
            data_collator=data_collator,
            compute_metrics=compute_metrics,
        )

args_r0 = build_training_args(os.path.join(BASE_PATH, 'models/jobbert_cert_round0'), epochs=5, lr=2e-5)
print("TrainingArguments for Round 0 configured successfully.")
````

### 2.5 Semi-Supervised Self-Training Pseudo-Labeling (Confidence Threshold tau = 0.85)


````python
def split_sentences(text, max_chars=800):
    sents = re.split(r'(?<=[.!?])\s+', str(text).strip())
    chunks, current = [], ''
    for s in sents:
        if len(current) + len(s) > max_chars and current:
            chunks.append(current.strip())
            current = s
        else:
            current = (current + ' ' + s).strip()
    if current:
        chunks.append(current)
    return chunks
````

````python
CONFIDENCE_THRESHOLD = 0.85
MAX_PSEUDO_SENTENCES = 5000

model_r0_inf = AutoModelForTokenClassification.from_pretrained(
    os.path.join(BASE_PATH, 'models/jobbert_cert_round0')
).to(device).eval()
print('Round 0 model loaded for pseudo-labeling')
````

````python
def get_word_level_predictions(sentence, tokenizer, model, id2label):
    words = sentence.split()
    if not words:
        return [], []
    inputs   = tokenizer(
        words,
        is_split_into_words=True,
        return_tensors='pt',
        truncation=True,
        max_length=128,
    )
    word_ids = inputs.word_ids(0)
    inputs   = {k: v.to(device) for k, v in inputs.items()}

    with torch.no_grad():
        logits = model(**inputs).logits
    probs = F.softmax(logits, dim=-1)[0]

    word_labels = []
    seen = set()
    for idx, wid in enumerate(word_ids):
        if wid is None or wid in seen:
            continue
        seen.add(wid)
        word_labels.append(id2label[probs[idx].argmax().item()])

    return words, word_labels[:len(words)]
````

````python
def pseudo_label_sentence(text):
    pseudo_records = []
    for sent in split_sentences(clean_raw(text)):
        words, token_labels = get_word_level_predictions(
            sent, tokenizer_cert, model_r0_inf, ID2LABEL_CERT
        )
        if len(words) < 3:
            continue

        words        = words[:len(token_labels)]
        token_labels = token_labels[:len(words)]

        inputs   = tokenizer_cert(
            words, is_split_into_words=True,
            return_tensors='pt', truncation=True, max_length=128,
        )
        word_ids = inputs.word_ids(0)
        inputs   = {k: v.to(device) for k, v in inputs.items()}
        with torch.no_grad():
            logits = model_r0_inf(**inputs).logits
        probs = F.softmax(logits, dim=-1)[0]

        token_confs = []
        seen = set()
        for idx, wid in enumerate(word_ids):
            if wid is None or wid in seen:
                continue
            seen.add(wid)
            token_confs.append(probs[idx].max().item())
        token_confs = token_confs[:len(words)]

        min_len      = min(len(words), len(token_labels), len(token_confs))
        words        = words[:min_len]
        token_labels = token_labels[:min_len]
        token_confs  = token_confs[:min_len]

        final_labels = ['O'] * min_len
        i = 0
        while i < min_len:
            if token_labels[i].startswith('B-'):
                span_idx = [i]
                j = i + 1
                while j < min_len and token_labels[j].startswith('I-'):
                    span_idx.append(j)
                    j += 1
                if min(token_confs[k] for k in span_idx) >= CONFIDENCE_THRESHOLD:
                    for k in span_idx:
                        final_labels[k] = token_labels[k]
                i = j
            else:
                i += 1

        if any(l != 'O' for l in final_labels):
            pseudo_records.append({'tokens': words, 'labels': final_labels})
    return pseudo_records
````

### 2.7 Penyatuan Inferensi Multi-Ontologi (SKILL, KNOWLEDGE, CERT)


````python
def extract_spans(words, labels):
    entities, current_span, current_label = [], [], None
    for word, label in zip(words, labels):
        if label.startswith('B-'):
            if current_span:
                entities.append((' '.join(current_span), current_label))
            current_span  = [word]
            current_label = label[2:]
        elif label.startswith('I-') and current_span:
            current_span.append(word)
        else:
            if current_span:
                entities.append((' '.join(current_span), current_label))
            current_span, current_label = [], None
    if current_span:
        entities.append((' '.join(current_span), current_label))
    return entities
````

````python
def merge_and_extract(sentence):
    _, labels_cert      = get_word_level_predictions(sentence, tokenizer_cert_inf, model_cert_inf, id2label_cert_inf)
    words, labels_skill = get_word_level_predictions(sentence, tokenizer_skill, model_skill, id2label_skill)
    _, labels_ke        = get_word_level_predictions(sentence, tokenizer_ke, model_ke, id2label_ke)

    length = min(len(labels_cert), len(labels_skill), len(labels_ke))
    if length == 0:
        return []

    merged = []
    for i in range(length):
        lc = labels_cert[i]
        ls = labels_skill[i]
        lk = labels_ke[i]

        if lc != 'O':
            merged.append(lc)
        elif ls in ('B', 'I'):
            merged.append('B-SKILL' if ls == 'B' else 'I-SKILL')
        elif lk in ('B', 'I'):
            merged.append('B-KNOWLEDGE' if lk == 'B' else 'I-KNOWLEDGE')
        else:
            merged.append('O')

    return extract_spans(words[:length], merged)
````

### 2.3 Metodologi Topologi Graf Kompetensi: Pointwise Mutual Information (PMI), PageRank, dan Betweenness Centrality

Hubungan asosiasi antar-keahlian dimodelkan menggunakan kerangka teori graf berbasis korelasi informasi mutual. Pengukuran ko-okurensi sederhana memiliki bias matematis terhadap istilah yang sangat umum muncul. Oleh karena itu, bobot keterhubungan (*edge weight*) dihitung menggunakan **Pointwise Mutual Information (PMI)**:
$$	ext{PMI}(x, y) = \log_2 rac{P(x, y)}{P(x) P(y)} = \log_2 \left( rac{C(x, y) \cdot N}{C(x) \cdot C(y)} ight)$$
di mana $C(x,y)$ adalah jumlah dokumen di mana keahlian $x$ dan $y$ muncul bersamaan, $C(x)$ dan $C(y)$ adalah frekuensi marjinal masing-masing keahlian, dan $N$ adalah total dokumen.

#### Konstruksi Graf Kompetensi:
Graf tak-berarah berbobot $G = (V, E, W)$ dibangun dengan kriteria penyaringan:
$$E = \left\{ (x, y) \;\middle|\; 	ext{PMI}(x, y) > 0 \quad 	ext{dan} \quad C(x, y) \ge 5 ight\}, \quad W(x, y) = 	ext{PMI}(x, y)$$

#### Formulasi Metrik Sentralitas Graf:
1. **PageRank Centrality (Keahlian Ubiquitous):**
   Mengukur probabilitas stasioner suatu keahlian dalam rantai Markov acak dengan probabilitas melompat (*damping factor*) $d = 0.85$:
   $$PR(u) = rac{1 - d}{|V|} + d \sum_{v \in \mathcal{N}_{	ext{in}}(u)} rac{PR(v)}{d_{	ext{out}}(v)}$$
   Keahlian dengan PageRank tinggi mencerminkan kompetensi *ubiquitous* yang menjadi fondasi umum di seluruh sektor industri.

2. **Betweenness Centrality (Bridge Skills / Keahlian Jembatan):**
   Mengukur proporsi jalur terpendek (*shortest paths*) antar seluruh pasangan simpul yang melewati simpul $v$:
   $$C_B(v) = \sum_{s 
e v 
e t \in V} rac{\sigma_{st}(v)}{\sigma_{st}}$$
   di mana $\sigma_{st}$ adalah total lintasan terpendek dari $s$ ke $t$, dan $\sigma_{st}(v)$ adalah lintasan yang melewati simpul $v$. Simpul dengan $C_B(v)$ tinggi dinamakan **Bridge Skills**—keahlian yang menjembatani klaster industri terisolasi.

3. **Greedy Modularity Community Detection (Clauset-Newman-Moore):**
   Mempartisi graf ke dalam komunitas kohesif dengan memaksimalkan fungsi modularitas $Q$:
   $$Q = rac{1}{2m} \sum_{i,j} \left[ A_{ij} - rac{k_i k_j}{2m} ight] \delta(c_i, c_j)$$


````python
GENERIC_SOFT_SKILLS = {
    'communication', 'written communication', 'verbal communication', 'interpersonal skills',
    'leadership', 'teamwork', 'problem solving', 'organizational skills', 'critical thinking',
    'attention to detail', 'work independently', 'bachelor', "bachelor's degree", 'degree',
    'experience', 'skills', 'ability', 'requirements', 'support', 'management',
    'collaborate', 'responsibilities', 'opportunity', 'customer service', 'time management'
}

df_ner_tech = df_ner[
    ~df_ner['entity_text'].str.lower().isin(GENERIC_SOFT_SKILLS) &
    (df_ner['entity_text'].str.len() >= 3)
].copy()

top40_entities = df_ner_tech['entity_text'].value_counts().head(40).index.tolist()
df_ner_top40 = df_ner_tech[df_ner_tech['entity_text'].isin(top40_entities)]

total_jobs = df['job_id'].nunique()
entity_job_counts = df_ner_top40.groupby('entity_text')['job_id'].nunique()
P_x = entity_job_counts / total_jobs

job_entities = df_ner_top40.groupby('job_id')['entity_text'].apply(set)

cooc_matrix = pd.DataFrame(0, index=top40_entities, columns=top40_entities)
for entity_set in job_entities:
    ents = [e for e in top40_entities if e in entity_set]
    for i in range(len(ents)):
        for j in range(i + 1, len(ents)):
            cooc_matrix.loc[ents[i], ents[j]] += 1
            cooc_matrix.loc[ents[j], ents[i]] += 1

P_xy = cooc_matrix / total_jobs

G_skill = nx.Graph()
entity_type_map = df_ner_top40.groupby('entity_text')['entity_label'].agg(lambda x: x.mode()[0]).to_dict()

for ent in top40_entities:
    G_skill.add_node(
        ent,
        entity_type=entity_type_map.get(ent, 'SKILL'),
        frequency=int(entity_job_counts.get(ent, 0))
    )

for i in range(len(top40_entities)):
    for j in range(i + 1, len(top40_entities)):
        e1, e2 = top40_entities[i], top40_entities[j]
        c = cooc_matrix.loc[e1, e2]
        if c >= 5:
            pmi = np.log2((P_xy.loc[e1, e2] + 1e-9) / (P_x[e1] * P_x[e2] + 1e-9))
            if pmi > 0.0:
                G_skill.add_edge(e1, e2, weight=float(pmi), count=int(c))

pagerank_scores  = nx.pagerank(G_skill, weight='weight')
betweenness_scores = nx.betweenness_centrality(G_skill, weight='weight', normalized=True)
degree_scores    = nx.degree_centrality(G_skill)

communities = list(nx_comm.greedy_modularity_communities(G_skill))
community_map = {}
for comm_id, comm in enumerate(communities):
    for node in comm:
        community_map[node] = comm_id

centrality_rows = []
for node in G_skill.nodes():
    centrality_rows.append({
        'entity_text': node,
        'entity_type': entity_type_map.get(node, 'SKILL'),
        'community_id': community_map.get(node, 0),
        'pagerank_centrality': round(pagerank_scores[node], 5),
        'betweenness_centrality': round(betweenness_scores[node], 5),
        'degree_centrality': round(degree_scores[node], 4),
        'job_frequency': int(entity_job_counts.get(node, 0))
    })

df_centrality = pd.DataFrame(centrality_rows).sort_values('betweenness_centrality', ascending=False)
df_bridge_skills = df_centrality.head(10).copy()

df_centrality.to_csv(os.path.join(BASE_PATH, 'exports/tables_csv/full_ner_centrality.csv'), index=False)
df_centrality.to_csv('exports/tables_csv/full_ner_centrality.csv', index=False)

df_bridge_skills.to_csv(os.path.join(BASE_PATH, 'exports/tables_csv/top_bridge_skills.csv'), index=False)
df_bridge_skills.to_csv('exports/tables_csv/top_bridge_skills.csv', index=False)

print("TOP 10 BRIDGE SKILLS (High Betweenness Centrality — Connecting Diverse Domains)")
print(df_bridge_skills[['entity_text', 'entity_type', 'betweenness_centrality', 'pagerank_centrality', 'community_id']].to_string(index=False))

plt.figure(figsize=(14, 10))
pos = nx.spring_layout(G_skill, k=0.55, seed=42)

node_colors = [community_map.get(node, 0) for node in G_skill.nodes()]
node_sizes  = [pagerank_scores[node] * 18000 + 300 for node in G_skill.nodes()]
edge_weights = [d['weight'] * 0.8 for _, _, d in G_skill.edges(data=True)]

nx.draw_networkx_nodes(G_skill, pos, node_size=node_sizes, node_color=node_colors, cmap='tab10', alpha=0.9, edgecolors='black', linewidths=1.2)
nx.draw_networkx_edges(G_skill, pos, width=edge_weights, alpha=0.35, edge_color='#7f8c8d')
nx.draw_networkx_labels(G_skill, pos, font_size=9, font_weight='bold', font_family='sans-serif')

plt.title('SkillGraph-AI: PMI Technical Competency Network & Topological Communities\n(Node Size: PageRank Centrality | Colors: Modularity Communities)', fontsize=13, fontweight='bold', pad=15)
plt.axis('off')
plt.tight_layout()
plt.savefig(os.path.join(BASE_PATH, 'figures/fig_ner_network.png'), dpi=300, bbox_inches='tight')
plt.savefig('exports/figures/fig_ner_network.png', dpi=300, bbox_inches='tight')
plt.show()

edge_x, edge_y = [], []
for edge in G_skill.edges():
    x0, y0 = pos[edge[0]]
    x1, y1 = pos[edge[1]]
    edge_x.extend([x0, x1, None])
    edge_y.extend([y0, y1, None])

edge_trace = go.Scatter(x=edge_x, y=edge_y, line=dict(width=0.8, color='#888'), hoverinfo='none', mode='lines')

node_x, node_y, node_text, node_color_list = [], [], [], []
for node in G_skill.nodes():
    x, y = pos[node]
    node_x.append(x)
    node_y.append(y)
    node_color_list.append(community_map.get(node, 0))
    node_text.append(
        f"<b>{node.upper()}</b> ({entity_type_map.get(node, 'SKILL')})<br>"
        f"Community: {community_map.get(node, 0)}<br>"
        f"Betweenness (Bridge): {betweenness_scores[node]:.4f}<br>"
        f"PageRank (Core): {pagerank_scores[node]:.4f}<br>"
        f"Postings: {entity_job_counts.get(node, 0):,}"
    )

node_trace = go.Scatter(
    x=node_x, y=node_y, mode='markers+text',
    text=[node for node in G_skill.nodes()],
    textposition="top center",
    hoverinfo='text',
    hovertext=node_text,
    marker=dict(
        showscale=True,
        colorscale='Viridis',
        color=node_color_list,
        size=[pagerank_scores[n] * 600 + 15 for n in G_skill.nodes()],
        line_width=1.5
    )
)

fig_plotly_network = go.Figure(
    data=[edge_trace, node_trace],
    layout=go.Layout(
        title="<b>Interactive SkillGraph-AI Knowledge Network</b><br>Hover for Bridge Skills & Centrality Topology",
        showlegend=False,
        hovermode='closest',
        margin=dict(b=20, l=5, r=5, t=50),
        xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
        yaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
        template='plotly_white'
    )
)
fig_plotly_network.write_html(os.path.join(BASE_PATH, 'exports/plotly_html/plotly_skill_network.html'))
fig_plotly_network.write_html('exports/plotly_html/plotly_skill_network.html')
print("Saved fig_ner_network.png (300 DPI) and interactive plotly_skill_network.html")
````

## 3. Task B: Pemetaan Taksonomi Okupasi Semantik Menggunakan Sentence-BERT, Reduksi Manifold UMAP, HDBSCAN, dan GMM

### 3.1 Landasan Teori Representasi Semantik Sentence-BERT dan Manifold UMAP
Pengelompokan dokumen lowongan kerja membutuhkan representasi vektor kontinu yang menangkap makna kontekstual kalimat secara utuh. Model leksikal tradisional (TF-IDF) rentan terhadap masalah sinonimi (*job title synonymy*) dan variasi frasa.

#### Arsitektur Sentence-BERT (`all-MiniLM-L6-v2`):
Pipeline menggunakan arsitektur jaringan siam (*Siamese network*) Sentence-BERT yang dioptimalkan dengan fungsi objektif *cosine similarity*. Vektor embedding dokumen 384 dimensi diperoleh melalui *mean pooling* berbobot terhadap seluruh token representasi subword:
$$\mathbf{u} = rac{\sum_{i=1}^{L} m_i \cdot \mathbf{h}_i}{\sum_{i=1}^{L} m_i}$$
di mana $\mathbf{h}_i$ adalah vektor tersembunyi token ke-$i$, $m_i \in \{0, 1\}$ adalah attention mask, dan $L$ adalah panjang sekuens. Untuk mengatasi teks yang panjang, diterapkan mekanisme *chunking* paragraf yang dirata-ratakan secara seragam.

#### Teori Reduksi Manifold Non-Linier UMAP:
Metode reduksi linear seperti PCA gagal menangkap struktur topologi data teks yang bersifat manifold non-linier pada dimensi tinggi. Task B menerapkan **Uniform Manifold Approximation and Projection (UMAP)** yang berakar pada geometri Riemannian dan topologi aljabar:
1. Mengonstruksi representasi kompleks simplisial fuzzy (*fuzzy simplicial set*) pada ruang berdimensi tinggi berdasarkan jarak metrik kosinus.
2. Mengoptimalkan posisi koordinat ruang proyeksi dimensi rendah dengan meminimalkan *fuzzy set cross-entropy*:
$$C(P, Q) = \sum_{e} \left[ p_e \log \left(rac{p_e}{q_e}ight) + (1 - p_e) \log \left(rac{1 - p_e}{1 - q_e}ight) ight]$$
* **Proyeksi 50 Dimensi (Intermediate 50-D):** Digunakan sebagai masukan klastering densitas, menghindari *curse of dimensionality* sekaligus mempertahankan 90%+ varians struktur lokal-global.
* **Proyeksi 2 Dimensi (Visual 2-D):** Digunakan secara khusus untuk pemetaan koordinat visualisasi sebaran galaksi okupasi.

### 3.1 Pembangkitan Representasi Vektor Kalimat Sentence-BERT (384-D)


````python
SBERT_MODEL_NAME = 'all-MiniLM-L6-v2'
EMBED_CLUSTER_PATH = os.path.join(BASE_PATH, 'data/processed/sbert_cluster_embeddings.npy')

CHUNK_WINDOW = 200
CHUNK_STRIDE = 150

def chunk_text(text, window=CHUNK_WINDOW, stride=CHUNK_STRIDE):
    words = str(text).split()
    if len(words) <= window:
        return [' '.join(words)] if words else ['']
    chunks = []
    for start in range(0, len(words), stride):
        chunk = words[start:start + window]
        if chunk:
            chunks.append(' '.join(chunk))
        if start + window >= len(words):
            break
    return chunks

if Path(EMBED_CLUSTER_PATH).exists():
    cluster_embeddings = np.load(EMBED_CLUSTER_PATH)
    print(f'Loaded cached embeddings: {cluster_embeddings.shape}')
else:
    print(f'Loading SBERT model: {SBERT_MODEL_NAME}')
    sbert_model = SentenceTransformer(SBERT_MODEL_NAME)

    all_chunks, owner_idx = [], []
    for i, txt in enumerate(df_cl['text_raw'].fillna('').tolist()):
        for ch in chunk_text(txt):
            all_chunks.append(ch)
            owner_idx.append(i)
    owner_idx = np.asarray(owner_idx)
    print(f'{len(df_cl):,} jobs -> {len(all_chunks):,} chunks '
          f'(avg {len(all_chunks)/len(df_cl):.2f} chunks/job)')

    chunk_vecs = sbert_model.encode(
        all_chunks,
        batch_size=64,
        show_progress_bar=True,
        convert_to_numpy=True,
        normalize_embeddings=False
    )

    dim = chunk_vecs.shape[1]
    cluster_embeddings = np.zeros((len(df_cl), dim), dtype=np.float32)
    np.add.at(cluster_embeddings, owner_idx, chunk_vecs)
    counts = np.bincount(owner_idx, minlength=len(df_cl)).reshape(-1, 1)
    cluster_embeddings /= np.maximum(counts, 1)

    norms = np.linalg.norm(cluster_embeddings, axis=1, keepdims=True)
    cluster_embeddings /= np.maximum(norms, 1e-12)

    np.save(EMBED_CLUSTER_PATH, cluster_embeddings)
    print(f'Pooled embeddings saved to {EMBED_CLUSTER_PATH}')

print(f'Embedding dimension: {cluster_embeddings.shape[1]}')
print(f'Total vectors: {cluster_embeddings.shape[0]:,}')
````

````python
hdbscan_path = os.path.join(BASE_PATH, 'models/hdbscan_model.joblib')

if not Path(hdbscan_path).exists():
    print('Running HDBSCAN on 50D UMAP embeddings...')
    hdb_model = hdbscan.HDBSCAN(
        min_cluster_size=50, min_samples=5,
        metric='euclidean', cluster_selection_method='eom',
        cluster_selection_epsilon=0.0, prediction_data=True,
        gen_min_span_tree=True, core_dist_n_jobs=-1
    )
    hdb_labels = hdb_model.fit_predict(emb_umap_hi)

    n_clusters_hdb = len(set(hdb_labels)) - (1 if -1 in hdb_labels else 0)
    n_noise_hdb    = int((hdb_labels == -1).sum())
    coverage_hdb   = (1 - n_noise_hdb / len(hdb_labels)) * 100

    print(f'Number of clusters: {n_clusters_hdb}')
    print(f'Coverage (non-noise): {coverage_hdb:.1f}%')
    print(f'Noise points: {n_noise_hdb:,} ({100 - coverage_hdb:.1f}%)')

    mask_valid_hdb = hdb_labels != -1
    if mask_valid_hdb.sum() > 500 and n_clusters_hdb >= 2:
        sample_sz = min(5000, int(mask_valid_hdb.sum()))
        sil_hdb  = silhouette_score(emb_umap_hi[mask_valid_hdb], hdb_labels[mask_valid_hdb],
                                     sample_size=sample_sz, random_state=42)
        dbi_hdb  = davies_bouldin_score(emb_umap_hi[mask_valid_hdb], hdb_labels[mask_valid_hdb])
        ch_hdb   = calinski_harabasz_score(emb_umap_hi[mask_valid_hdb], hdb_labels[mask_valid_hdb])
        dbcv_hdb = float(hdb_model.relative_validity_)
        print(f'Silhouette Score: {sil_hdb:.4f}')
        print(f'Davies-Bouldin Score: {dbi_hdb:.4f}')
        print(f'Calinski-Harabasz Score: {ch_hdb:.2f}')
        print(f'DBCV Relative Validity: {dbcv_hdb:.4f}')
    else:
        sil_hdb, dbi_hdb, ch_hdb, dbcv_hdb = 0.0, float('inf'), 0.0, 0.0
        print('Not enough clustered points to compute metrics')

    df_cl['hdbscan_label'] = hdb_labels
    joblib.dump(hdb_model, hdbscan_path)
    print('HDBSCAN model saved to Drive')
else:
    hdb_model  = joblib.load(hdbscan_path)
    hdb_labels = hdb_model.labels_
    df_cl['hdbscan_label'] = hdb_labels
    n_clusters_hdb = len(set(hdb_labels)) - (1 if -1 in hdb_labels else 0)
    n_noise_hdb    = int((hdb_labels == -1).sum())
    coverage_hdb   = (1 - n_noise_hdb / len(hdb_labels)) * 100

    mask_valid_hdb = hdb_labels != -1
    if mask_valid_hdb.sum() > 500 and n_clusters_hdb >= 2:
        sample_sz = min(5000, int(mask_valid_hdb.sum()))
        sil_hdb  = silhouette_score(emb_umap_hi[mask_valid_hdb], hdb_labels[mask_valid_hdb],
                                     sample_size=sample_sz, random_state=42)
        dbi_hdb  = davies_bouldin_score(emb_umap_hi[mask_valid_hdb], hdb_labels[mask_valid_hdb])
        ch_hdb   = calinski_harabasz_score(emb_umap_hi[mask_valid_hdb], hdb_labels[mask_valid_hdb])
        dbcv_hdb = float(hdb_model.relative_validity_)
    else:
        sil_hdb, dbi_hdb, ch_hdb, dbcv_hdb = 0.0, float('inf'), 0.0, 0.0
    print(f'HDBSCAN loaded from Drive — {n_clusters_hdb} clusters')
````

### 3.4 Klastering Probabilistik Gaussian Mixture Model (GMM)


````python
gmm_path = os.path.join(BASE_PATH, 'models/gmm_model.joblib')

if not Path(gmm_path).exists():
    n_components_range = range(5, 51, 3)
    COV_TYPE  = 'full'
    REG_COVAR = 1e-3

    print(f'GMM search | cov={COV_TYPE} | range=5-50 | reg_covar={REG_COVAR}')
    gmm_bic_results = []
    for n_comp in tqdm(n_components_range, desc='GMM BIC/AIC search'):
        try:
            gmm_temp = GaussianMixture(
                n_components=n_comp, covariance_type=COV_TYPE,
                random_state=42, max_iter=300, n_init=3, reg_covar=REG_COVAR
            )
            gmm_temp.fit(emb_umap_hi)
            bic_val = gmm_temp.bic(emb_umap_hi)
            aic_val = gmm_temp.aic(emb_umap_hi)
            gmm_bic_results.append({'n': n_comp, 'bic': bic_val, 'aic': aic_val})
            print(f'n={n_comp:2d} | BIC: {bic_val:>15,.0f} | AIC: {aic_val:>15,.0f}')
        except (ValueError, np.linalg.LinAlgError) as e:
            print(f'n={n_comp:2d} FAILED -> {e}')

    df_bic = pd.DataFrame(gmm_bic_results)
    best_n_gmm = int(df_bic.loc[df_bic['bic'].idxmin(), 'n'])
    df_bic['bic_gain_pct'] = (-df_bic['bic'].diff() / df_bic['bic'].abs().shift()) * 100
    elbow_candidates = df_bic[df_bic['bic_gain_pct'] < 1.0]['n'].tolist()
    elbow_n = elbow_candidates[0] if elbow_candidates else best_n_gmm

    df_bic.to_csv(os.path.join(BASE_PATH, 'data/processed/gmm_bic_results.csv'), index=False)

    while best_n_gmm >= 2:
        try:
            gmm_final = GaussianMixture(
                n_components=best_n_gmm, covariance_type=COV_TYPE,
                random_state=42, max_iter=300, n_init=10,
                reg_covar=max(REG_COVAR, 1e-4), verbose=0
            )
            gmm_final.fit(emb_umap_hi)
            gmm_labels     = gmm_final.predict(emb_umap_hi)
            gmm_probs      = gmm_final.predict_proba(emb_umap_hi)
            gmm_confidence = gmm_probs.max(axis=1)
            sil_gmm = silhouette_score(emb_umap_hi, gmm_labels,
                                       sample_size=min(5000, len(emb_umap_hi)), random_state=42)
            dbi_gmm = davies_bouldin_score(emb_umap_hi, gmm_labels)
            ch_gmm  = calinski_harabasz_score(emb_umap_hi, gmm_labels)
            bic_final = gmm_final.bic(emb_umap_hi)
            aic_final = gmm_final.aic(emb_umap_hi)
            df_cl['gmm_label']      = gmm_labels
            df_cl['gmm_confidence'] = gmm_confidence
            joblib.dump(gmm_final, gmm_path)
            print(f'GMM saved to Drive — {best_n_gmm} components')
            break
        except (ValueError, np.linalg.LinAlgError):
            print(f'GMM failed at n={best_n_gmm}, retrying lower...')
            best_n_gmm -= 1
else:
    gmm_final      = joblib.load(gmm_path)
    gmm_labels     = gmm_final.predict(emb_umap_hi)
    gmm_probs      = gmm_final.predict_proba(emb_umap_hi)
    gmm_confidence = gmm_probs.max(axis=1)
    best_n_gmm     = gmm_final.n_components
    df_cl['gmm_label']      = gmm_labels
    df_cl['gmm_confidence'] = gmm_confidence
    sil_gmm = silhouette_score(emb_umap_hi, gmm_labels,
                               sample_size=min(5000, len(emb_umap_hi)), random_state=42)
    dbi_gmm = davies_bouldin_score(emb_umap_hi, gmm_labels)
    ch_gmm  = calinski_harabasz_score(emb_umap_hi, gmm_labels)
    bic_final = gmm_final.bic(emb_umap_hi)
    aic_final = gmm_final.aic(emb_umap_hi)

    df_bic = pd.read_csv(os.path.join(BASE_PATH, 'data/processed/gmm_bic_results.csv'))
    elbow_candidates = df_bic[df_bic['bic_gain_pct'] < 1.0]['n'].tolist()
    elbow_n = elbow_candidates[0] if elbow_candidates else best_n_gmm
    print(f'GMM loaded from Drive — {best_n_gmm} components')
````

#### Interpretasi Analitis: Optimasi Komponen GMM Berdasarkan Bayesian Information Criterion (BIC)
Pencarian jumlah komponen optimal pada model Gaussian Mixture Model (GMM) dengan matriks kovarians penuh (*full covariance*) menghasilkan kurva seleksi model sebagai berikut:
1. **Penurunan Nilai BIC yang Signifikan:** Kurva BIC menunjukkan penurunan yang sangat tajam seiring peningkatan $k$ dari 5 menuju 8, dan mencapai titik lembah optimal pada **$k = 8$ komponen** dengan nilai BIC terendah sebesar **-2.019.337** (AIC: **-2.095.817**).
2. **Keseimbangan Kompleksitas Model (Occam's Razor):** Pemilihan $k = 8$ komponen makro menjamin pengelompokan keluarga okupasi yang parsimonius (*parsimonious clustering*), mencegah terjadinya partisi mikro yang berlebihan (*over-clustering*), serta memberikan ruang probabilitas posterior yang stabil untuk mendekomposisi titik derau.

### 3.2 Inovasi Metodologi: Two-Stage HDBSCAN-GMM Cascade Clustering Berbasis Entropi Shannon

Sebagaimana telah dibuktikan, HDBSCAN meninggalkan 33,4% data (3.341 lowongan) sebagai derau. Membuang data ini akan menghilangkan sepertiga dinamika pasar tenaga kerja.

Untuk mengatasi limitasi tersebut, dirancang inovasi **Two-Stage Cascade Clustering** yang memanfaatkan **Entropi Shannon** dari probabilitas posterior GMM:
$$\mathcal{H}(x_i) = -\sum_{k=1}^K p(z_k | x_i) \log_2 p(z_k | x_i)$$
di mana $p(z_k | x_i)$ adalah probabilitas keanggotaan posterior lowongan $x_i$ pada komponen GMM ke-$k$.

```
[Seluruh 10.000 Lowongan Kerja]
       │
       ▼ (Stage 1: HDBSCAN Clustering)
       ├─────────────────────────────────────────┐
       ▼                                         ▼
[Core Dense Clusters]                     [HDBSCAN Noise (-1)]
(6.659 lowongan, 66.6%)                   (3.341 lowongan, 33.4%)
                                                 │
                                                 ▼ (Stage 2: GMM Posterior & Shannon Entropy)
                                  ┌──────────────┴──────────────┐
                                  ▼                             ▼
                      [Low Entropy / High Conf]     [High Entropy (H ≥ 0.50)]
                      (p_max ≥ 0.85)                            │
                      Niche / Ultra-Specialized                 ├────────────────────────┐
                      (3.319 lowongan)                          ▼                        ▼
                                                     [High Entities ≥ 10]     [Low Entities ≤ 4]
                                                     Hybrid / Cross-Func      Generic Staffing
                                                     (14 lowongan)            (8 lowongan)
```


````python
entropy_all = -np.sum(gmm_probs * np.log(gmm_probs + 1e-12), axis=1)
df_cl['cluster_entropy'] = entropy_all
df_cl['gmm_max_confidence'] = gmm_confidence

cols_to_merge_cl = [c for c in ['ner_knowledge_count', 'ner_skill_count'] if c in df.columns]
for c in cols_to_merge_cl:
    if c in df_cl.columns:
        df_cl = df_cl.drop(columns=[c])

if cols_to_merge_cl:
    df_cl = df_cl.merge(df[['job_id'] + cols_to_merge_cl], on='job_id', how='left')
else:
    df_cl['ner_knowledge_count'] = 0
    df_cl['ner_skill_count'] = 0

noise_mask = df_cl['hdbscan_label'] == -1
median_noise_entropy = df_cl.loc[noise_mask, 'cluster_entropy'].median()
median_knowledge_cnt = df_cl['ner_knowledge_count'].median()

cascade_categories = []
is_hybrid_role_flags = []

for idx, row in df_cl.iterrows():
    if row['hdbscan_label'] != -1:
        cascade_categories.append('Core Dense Cluster')
        is_hybrid_role_flags.append(0)
    else:
        conf_val = row['gmm_max_confidence']
        ent_val  = row['cluster_entropy']
        know_val = row.get('ner_knowledge_count', 0)

        if conf_val < 0.60 and know_val >= median_knowledge_cnt:
            cascade_categories.append('Hybrid / Cross-Functional')
            is_hybrid_role_flags.append(1)
        elif conf_val >= 0.60:
            cascade_categories.append('Niche / Ultra-Specialized')
            is_hybrid_role_flags.append(0)
        else:
            cascade_categories.append('Generic Staffing / Boilerplate')
            is_hybrid_role_flags.append(0)

df_cl['cascade_category'] = cascade_categories
df_cl['is_hybrid_role']   = is_hybrid_role_flags

cascade_summary = df_cl.groupby('cascade_category').agg(
    total_jobs=('job_id', 'count'),
    mean_salary=('salary_mid', 'mean'),
    median_salary=('salary_mid', 'median'),
    mean_entropy=('cluster_entropy', 'mean'),
    mean_confidence=('gmm_max_confidence', 'mean'),
    mean_knowledge_entities=('ner_knowledge_count', 'mean')
).reset_index()

cascade_summary.to_csv(os.path.join(BASE_PATH, 'exports/tables_csv/noise_cascade_summary.csv'), index=False)
cascade_summary.to_csv('exports/tables_csv/noise_cascade_summary.csv', index=False)

print('TWO-STAGE HDBSCAN-GMM CASCADE NOISE DECOMPOSITION SUMMARY')
print(cascade_summary.to_string(index=False))

fig, axes = plt.subplots(1, 3, figsize=(20, 5.5))

noise_counts = df_cl[noise_mask]['cascade_category'].value_counts()
axes[0].bar(noise_counts.index, noise_counts.values, color=['#9b59b6', '#e67e22', '#95a5a6'], edgecolor='black', alpha=0.85)
axes[0].set_title(f'A. HDBSCAN Noise Decomposition (n = {noise_mask.sum():,} postings)', fontweight='bold')
axes[0].set_ylabel('Number of Jobs')
axes[0].tick_params(axis='x', rotation=15)
for i, v in enumerate(noise_counts.values):
    axes[0].text(i, v + 40, f'{v:,}\n({v/noise_mask.sum()*100:.1f}%)', ha='center', fontweight='bold', fontsize=9)

df_sal_cascade = df_cl.dropna(subset=['salary_mid']).copy()
sns.boxplot(
    data=df_sal_cascade, x='cascade_category', y='salary_mid',
    palette=['#2980b9', '#9b59b6', '#e67e22', '#95a5a6'], ax=axes[1],
    showmeans=True, meanprops={"marker":"o", "markerfacecolor":"red", "markeredgecolor":"red"}
)
axes[1].set_title('B. Annual Salary Comparison across Cascade Roles', fontweight='bold')
axes[1].set_ylabel('Annual Salary (USD/yr)')
axes[1].set_xlabel('')
axes[1].yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x*1e-3:.0f}K'))
axes[1].tick_params(axis='x', rotation=15)

sample_scatter = df_cl.sample(min(2500, len(df_cl)), random_state=42)
sns.scatterplot(
    data=sample_scatter, x='gmm_max_confidence', y='cluster_entropy', hue='cascade_category',
    palette=['#2980b9', '#9b59b6', '#e67e22', '#95a5a6'], alpha=0.6, s=25, ax=axes[2]
)
axes[2].set_title('C. GMM Confidence vs Shannon Entropy by Sub-Category', fontweight='bold')
axes[2].set_xlabel('Max GMM Component Confidence')
axes[2].set_ylabel('Shannon Entropy')
axes[2].legend(title='Cascade Subtype', fontsize=8, loc='upper right')

plt.suptitle('SkillGraph-AI: Two-Stage HDBSCAN-GMM Cascade Clustering Decomposition', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig(os.path.join(BASE_PATH, 'figures/fig_noise_cascade.png'), dpi=300, bbox_inches='tight')
plt.savefig('exports/figures/fig_noise_cascade.png', dpi=300, bbox_inches='tight')
plt.show()

fig_plotly_cascade = px.scatter(
    df_cl.sample(min(5000, len(df_cl)), random_state=42),
    x='umap_x', y='umap_y', color='cascade_category',
    hover_data=['title', 'hdbscan_label', 'gmm_label', 'cluster_entropy'],
    color_discrete_map={
        'Core Dense Cluster': '#2980b9',
        'Hybrid / Cross-Functional': '#9b59b6',
        'Niche / Ultra-Specialized': '#e67e22',
        'Generic Staffing / Boilerplate': '#bdc3c7'
    },
    title='<b>SkillGraph-AI Job Galaxy: Cascade Cluster & Noise Decomposition (UMAP 2D)</b>',
    template='plotly_white'
)
fig_plotly_cascade.write_html(os.path.join(BASE_PATH, 'exports/plotly_html/plotly_job_galaxy_cascade.html'))
fig_plotly_cascade.write_html('exports/plotly_html/plotly_job_galaxy_cascade.html')
print('Saved fig_noise_cascade.png (300 DPI) and interactive plotly_job_galaxy_cascade.html')

df_cl.to_parquet(os.path.join(BASE_PATH, 'exports/tables_csv/cluster_cascade_results.parquet'), index=False)
df_cl.to_parquet('exports/tables_csv/cluster_cascade_results.parquet', index=False)

df['hdbscan_cluster_id'] = df['job_id'].map(df_cl.set_index('job_id')['hdbscan_label']).fillna(-1).astype(int)
df['gmm_cluster_id']     = df['job_id'].map(df_cl.set_index('job_id')['gmm_label']).fillna(0).astype(int)
df['cluster_entropy']    = df['job_id'].map(df_cl.set_index('job_id')['cluster_entropy']).fillna(median_noise_entropy).astype(np.float32)
df['gmm_max_confidence'] = df['job_id'].map(df_cl.set_index('job_id')['gmm_max_confidence']).fillna(0.5).astype(np.float32)
df['is_hybrid_role']     = df['job_id'].map(df_cl.set_index('job_id')['is_hybrid_role']).fillna(0).astype(int)
df['cascade_category']   = df['job_id'].map(df_cl.set_index('job_id')['cascade_category']).fillna('Core Dense Cluster')

print('Propagated Task B cluster features to main dataframe successfully.')
````

### 4.1 Landasan Teori Fusi Fitur Sekuensial Multimodal (614 Dimensi)
Kelemahan umum dalam pemodelan prediksi upah konvensional adalah fragmentasi fitur: model hanya memanfaatkan metadata tabular dasar atau teks lepas secara terpisah. Task C menghadirkan inovasi **True Sequential Feature Fusion**, di mana seluruh representasi pengetahuan yang telah dipelajari dari Task A (entitas keahlian dan graf) dan Task B (klaster okupasi dan entropi cascade) disatukan ke dalam matriks berdimensi tinggi (**614 Dimensi**):

$$\mathbf{x}_{	ext{fused}} = \left[ \mathbf{x}_{	ext{tabular}} \,\|\, \mathbf{x}_{	ext{task\_a}} \,\|\, \mathbf{x}_{	ext{task\_b}} \,\|\, \mathbf{e}_{	ext{sbert}} \,\|\, \mathbf{v}_{	ext{tfidf\_svd}} ight] \in \mathbb{R}^{614}$$

2. **Representasi Semantik Kontekstual SBERT (384 Dimensi):** Embedding kontinu deskripsi peran dari model Sentence-BERT.
3. **Representasi Leksikal Domain TF-IDF TruncatedSVD (150 Dimensi):** Komponen leksikal reduksi SVD yang menangkap kata-kata teknis langka (*rare technical tokens*).

### 4.2 Skema Pembagian Dataset dan Target Encoding Bayesian Bebas Kebocoran Data (Zero Data Leakage)
Untuk memastikan validitas ilmiah berstandar internasional, seluruh proses evaluasi dijalankan dengan protokol ketat:
* **Stratified Splitting:** Dataset dibagi terstratifikasi berdasarkan kuintil kompensasi ke dalam **Train Set (70% / 6.807 baris)**, **Validation Set (15% / 1.459 baris)**, dan **Held-out Test Set (15% / 1.459 baris)**.
* **Target Encoding Bayesian Terregularisasi:**
  Variabel berkarstisitas tinggi (*high-cardinality nominals*) seperti `location`, `industry`, dan `company_name` dienkode menggunakan estimasi Bayesian dengan parameter penghalusan (*smoothing* $m = 10$):
  $$S_i = \lambda_i ar{y}_i + (1 - \lambda_i) ar{y}_{	ext{global}}, \quad \lambda_i = rac{n_i}{n_i + m}$$
  Encoder hanya di-*fit* pada Train Set dan diterapkan (*transform-only*) pada Validation dan Test Set, menjamin **Zero Data Leakage**.

### 4.3 Arsitektur Stacking Ensemble Regressor
Model prediksi kompensasi dibangun menggunakan kerangka kerja ansambel berjenjang (*Stacked Generalization*):
* **Level-0 Base Learners:**
  1. **XGBoost Regressor:** Pohon keputusan tergradien yang dioptimalkan melalui Optuna (40 *trials*, tuning `max_depth`, `learning_rate`, `subsample`, `colsample_bytree`).
  2. **LightGBM Regressor:** Algoritma berbasis histogram yang dioptimalkan melalui Optuna (30 *trials*, tuning `num_leaves`, `min_child_samples`, `feature_fraction`).
* **Level-1 Meta-Learner (RidgeCV):**
  Menggunakan model regresi linier terregularisasi L2 (*Ridge Regression*) dengan *cross-validation* internal untuk menentukan bobot linear optimal terhadap prediksi out-of-fold dari base learners, meminimalkan risiko *overfitting* meta-model.

### 4.2 Imputasi Terarah Jenjang Senioritas dari Judul Pekerjaan


````python
TITLE_SENIORITY_MAP = [
    (r"\b(intern|internship|co.?op)\b",                                   0),
    (r"\b(entry.?level|junior|jr\.?|graduate|trainee)\b",                1),
    (r"\b(associate|mid.?level|ii)\b",                                    2),
    (r"\b(senior|sr\.?|staff|iii|iv)\b",                                  3),
    (r"\b(lead|principal|manager|head|supervisor|superintendent)\b",      4),
    (r"\b(director|vp|vice.?president|chief|cto|cfo|ceo|president|"
     r"managing.?director|partner|dean|physician|dentist|surgeon|"
     r"attorney|counsel|lawyer)\b",                                        5),
]

def infer_seniority_from_title(title):
    t = str(title).lower()
    result = -1
    for pattern, level in TITLE_SENIORITY_MAP:
        if re.search(pattern, t):
            result = level
    return result

df_reg["seniority_from_title"] = df_reg["title"].apply(infer_seniority_from_title)

unknown_mask = df_reg["seniority_encoded"] == -1
recovered    = unknown_mask & (df_reg["seniority_from_title"] != -1)
df_reg.loc[recovered, "seniority_encoded"] = df_reg.loc[recovered, "seniority_from_title"]

still_unknown = df_reg["seniority_encoded"] == -1
industry_seniority_median = (
    df_reg[~still_unknown]
    .groupby("industry")["seniority_encoded"]
    .median()
)
global_median = df_reg.loc[~still_unknown, "seniority_encoded"].median()

def impute_by_industry(row):
    if row["seniority_encoded"] != -1:
        return row["seniority_encoded"]
    return industry_seniority_median.get(row["industry"], global_median)

df_reg["seniority_encoded"] = df_reg.apply(impute_by_industry, axis=1)
df_reg["seniority_encoded"] = df_reg["seniority_encoded"].round().astype(int)

print(f"Recovered from title : {recovered.sum():,} rows")
print(f"Remaining -1         : {(df_reg['seniority_encoded'] == -1).sum()}")
print(df_reg.groupby("seniority_encoded")["salary_mid"].agg(["mean", "count"]).round(0))
````

### 4.3 Pemuatan Representasi Semantik SBERT (384-D)


**Note on 300 Characters Limit:**<br>
The first 300 characters of the job description are extracted for SBERT embeddings because job postings consistently front-load the most critical information (role summaries and core objectives) at the very beginning. Furthermore, dense Transformer models have maximum token limits; isolating the introductory context maximizes the signal-to-noise ratio and prevents the embeddings from being diluted by boilerplate requirements typically found at the bottom of postings.


````python
REG_EMBED_PATH = os.path.join(BASE_PATH, "data/processed/sbert_regression_embeddings.npy")

if Path(REG_EMBED_PATH).exists():
    sbert_embeddings = np.load(REG_EMBED_PATH)
    print(f"Loaded cached SBERT embeddings: {sbert_embeddings.shape}")
else:
    if "sbert_model" not in dir():
        sbert_model = SentenceTransformer("all-MiniLM-L6-v2")
    context_texts = (
        df_reg["title"].fillna("") + " - " +
        df_reg["text_raw"].fillna("").str[:300]
    )
    sbert_embeddings = sbert_model.encode(
        context_texts.tolist(),
        batch_size=128,
        show_progress_bar=True,
        convert_to_numpy=True,
        normalize_embeddings=True,
    )
    np.save(REG_EMBED_PATH, sbert_embeddings)
    print(f"SBERT embeddings saved: {sbert_embeddings.shape}")
````

### 4.4 Pemuatan Komponen Leksikal TF-IDF TruncatedSVD (150-D)


````python
TFIDF_EMBED_PATH = os.path.join(BASE_PATH, "data/processed/tfidf_svd_embeddings.npy")

y_temp_for_split = np.log1p(df_reg["salary_mid"].values)
idx_all = np.arange(len(df_reg))
idx_train_temp, _ = train_test_split(idx_all, test_size=0.30, random_state=42)

if Path(TFIDF_EMBED_PATH).exists():
    tfidf_embeddings = np.load(TFIDF_EMBED_PATH)
    print(f"Loaded cached TF-IDF SVD embeddings: {tfidf_embeddings.shape}")
else:
    corpus = df_reg["text_raw"].fillna("").tolist()
    tfidf = TfidfVectorizer(
        max_features=15000,
        ngram_range=(1, 2),
        min_df=3,
        max_df=0.85,
        sublinear_tf=True,
    )
    tfidf.fit([corpus[i] for i in idx_train_temp])
    X_tfidf_sparse = tfidf.transform(corpus)
    svd = TruncatedSVD(n_components=150, random_state=42)
    svd.fit(X_tfidf_sparse[idx_train_temp])
    tfidf_embeddings = svd.transform(X_tfidf_sparse).astype(np.float32)
    np.save(TFIDF_EMBED_PATH, tfidf_embeddings)
    joblib.dump(tfidf, os.path.join(BASE_PATH, "models/tfidf_vectorizer.joblib"))
    joblib.dump(svd, os.path.join(BASE_PATH, "models/tfidf_svd.joblib"))
    print(f"TF-IDF SVD embeddings saved: {tfidf_embeddings.shape}")
````

### 4.5 Rekayasa Fitur Terstruktur dan Interaksi Domain (80-D)


````python
KEYWORDS = [
    "python", "sql", "aws", "azure", "gcp", "java", "javascript", "typescript",
    "react", "docker", "kubernetes", "spark", "machine learning", "deep learning",
    "data", "manager", "director", "lead", "senior", "principal", "architect",
    "agile", "scrum", "tableau", "excel", "r programming", "tensorflow", "pytorch",
    "compliance", "equity", "hedge fund", "private equity", "investment banking",
    "c++", "scala", "golang", "security", "devops", "mlops",
    "product manager", "data scientist", "data engineer", "software engineer",
    "quantitative", "physician", "dentist", "attorney", "surgeon", "pharmacist",
    "portfolio", "actuar", "underwriter", "cloud", "blockchain", "embedded",
]

def extract_keywords(text):
    t = str(text).lower()
    return [1 if kw in t else 0 for kw in KEYWORDS]

kw_flags = np.array(df_reg["text_raw"].fillna("").apply(extract_keywords).tolist())
kw_cols  = [f"kw_{kw.replace(' ', '_').replace('+', 'plus').replace('-', '_')}" for kw in KEYWORDS]
for i, col in enumerate(kw_cols):
    df_reg[col] = kw_flags[:, i]

df_reg["company_size_enc"] = pd.to_numeric(df_reg["company_size"], errors="coerce")
df_reg["company_size_enc"] = df_reg["company_size_enc"].fillna(df_reg["company_size_enc"].median())

for col in ["location", "industry"]:
    if col in df_reg.columns:
        df_reg[col] = df_reg[col].fillna("Unknown")

print(f"Keyword flags     : {len(kw_cols)}")
print(f"company_size NaN  : {df_reg['company_size_enc'].isna().sum()}")
````

### 4.6 Pembagian Stratifikasi dan Target Encoding Bayesian Bebas Kebocoran Data (Zero Data Leakage)


````python
te_path = os.path.join(BASE_PATH, 'models/target_encoder.joblib')

inherit_cols = [
    'ner_skill_count', 'ner_knowledge_count', 'ner_cert_count', 'has_certification', 'hard_to_soft_skill_ratio',
    'hdbscan_cluster_id', 'gmm_cluster_id', 'cluster_entropy', 'gmm_max_confidence', 'is_hybrid_role'
]

for col in inherit_cols:
    if col in df.columns:
        df_reg[col] = df_reg['job_id'].map(df.set_index('job_id')[col]).fillna(0)

y   = np.log1p(df_reg["salary_mid"].values)
idx = np.arange(len(df_reg))

salary_quintile = pd.qcut(df_reg["salary_mid"], q=5, labels=False)

idx_train, idx_temp, y_train, y_temp = train_test_split(
    idx, y, test_size=0.30, random_state=42, stratify=salary_quintile
)
salary_quintile_temp = salary_quintile.iloc[idx_temp].values
idx_val, idx_test, y_val, y_test = train_test_split(
    idx_temp, y_temp, test_size=0.50, random_state=42, stratify=salary_quintile_temp
)

idx_train_val = np.concatenate([idx_train, idx_val])
print(f'Dataset split: Train = {len(idx_train):,} | Val = {len(idx_val):,} | Test = {len(idx_test):,}')

te_cols = [c for c in ["location", "industry", "hdbscan_cluster_id", "gmm_cluster_id"] if c in df_reg.columns]

if not Path(te_path).exists():
    te = ce.TargetEncoder(cols=te_cols, smoothing=20)
    te.fit(df_reg.loc[idx_train, te_cols], y_train)
    joblib.dump(te, te_path)
    print('TargetEncoder fitted on idx_train and saved to Drive')
else:
    te = joblib.load(te_path)
    print('TargetEncoder loaded from Drive')

df_reg_enc = df_reg.copy()
df_reg_enc[[c + "_te" for c in te_cols]] = te.transform(df_reg[te_cols])
df_reg_enc = pd.get_dummies(df_reg_enc, columns=["formatted_work_type"], drop_first=True, dtype=int)
work_type_cols = [c for c in df_reg_enc.columns if c.startswith("formatted_work_type_")]

df_reg_enc["interact_sen_loc"] = df_reg_enc["seniority_encoded"] * df_reg_enc["location_te"]
df_reg_enc["interact_sen_exp"] = df_reg_enc["seniority_encoded"] * df_reg_enc["years_exp_req"]
df_reg_enc["interact_exp_ind"] = df_reg_enc["years_exp_req"]    * df_reg_enc["industry_te"]
df_reg_enc["seniority_x_cluster_te"] = df_reg_enc["seniority_encoded"] * df_reg_enc["gmm_cluster_id_te"]
interaction_cols = ["interact_sen_loc", "interact_sen_exp", "interact_exp_ind", "seniority_x_cluster_te"]

num_cols_base1 = (
    ["years_exp_req", "education_level", "seniority_encoded", "is_remote", "company_size_enc", "location_te", "industry_te"]
    + work_type_cols + kw_cols + ["interact_sen_loc", "interact_sen_exp", "interact_exp_ind"]
)

domain_features_task_ab = [
    "ner_skill_count", "ner_knowledge_count", "ner_cert_count", "has_certification", "hard_to_soft_skill_ratio",
    "hdbscan_cluster_id_te", "gmm_cluster_id_te", "cluster_entropy", "is_hybrid_role", "seniority_x_cluster_te"
]

num_cols_proposed = num_cols_base1 + [c for c in domain_features_task_ab if c not in num_cols_base1]

feat_names_proposed = (
    [f"sbert_{i}" for i in range(sbert_embeddings.shape[1])]
    + [f"tfidf_{i}" for i in range(tfidf_embeddings.shape[1])]
    + num_cols_proposed
)

X_num_base1    = df_reg_enc[num_cols_base1].fillna(0).values.astype(np.float32)
X_num_proposed = df_reg_enc[num_cols_proposed].fillna(0).values.astype(np.float32)

X_base1    = X_num_base1
X_base2    = np.hstack([sbert_embeddings, tfidf_embeddings, X_num_base1])
X_proposed = np.hstack([sbert_embeddings, tfidf_embeddings, X_num_proposed])

X_train_raw     = X_proposed[idx_train]
X_val_raw       = X_proposed[idx_val]
X_test_raw      = X_proposed[idx_test]
X_train_val_raw = np.vstack([X_train_raw, X_val_raw])
y_train_val     = np.concatenate([y_train, y_val])

print(f'Baseline 1 Feature Dimensions (Structured Only) : {X_base1.shape[1]}')
print(f'Baseline 2 Feature Dimensions (Structured + Text): {X_base2.shape[1]}')
print(f'Proposed SkillGraph-AI Full Feature Dimensions   : {X_proposed.shape[1]}')
````

### 4.7 Pelatihan Model Regresi Ansambel: XGBoost, LightGBM, dan Stacking


````python
xgb_path = os.path.join(BASE_PATH, 'models/xgb.joblib')

if not Path(xgb_path).exists():
    print("XGBoost Optuna Tuning (40 trials)")

    def objective_xgb(trial):
        params = {
            "n_estimators":      trial.suggest_int("n_estimators", 300, 1000, step=50),
            "max_depth":         trial.suggest_int("max_depth", 3, 8),
            "learning_rate":     trial.suggest_float("learning_rate", 0.005, 0.15, log=True),
            "subsample":         trial.suggest_float("subsample", 0.6, 1.0),
            "colsample_bytree":  trial.suggest_float("colsample_bytree", 0.4, 0.9),
            "colsample_bylevel": trial.suggest_float("colsample_bylevel", 0.4, 0.9),
            "min_child_weight":  trial.suggest_int("min_child_weight", 3, 20),
            "reg_alpha":         trial.suggest_float("reg_alpha", 1e-3, 5.0, log=True),
            "reg_lambda":        trial.suggest_float("reg_lambda", 1e-3, 5.0, log=True),
            "random_state": 42, "n_jobs": -1, "tree_method": "hist",
        }
        model = xgb.XGBRegressor(**params)
        model.fit(X_train_raw, y_train, eval_set=[(X_val_raw, y_val)], verbose=False)
        return np.sqrt(mean_squared_error(y_val, model.predict(X_val_raw)))

    study_xgb = optuna.create_study(direction="minimize")
    study_xgb.optimize(objective_xgb, n_trials=40)
    print(f"Best XGBoost val RMSE (log): {study_xgb.best_value:.5f}")

    best_xgb = xgb.XGBRegressor(
        **study_xgb.best_params, random_state=42, n_jobs=-1, tree_method="hist"
    )
    best_xgb.fit(X_train_val_raw, y_train_val, verbose=False)
    joblib.dump(best_xgb, xgb_path)
    print("XGBoost saved to Drive")
else:
    best_xgb = joblib.load(xgb_path)
    print("XGBoost loaded from Drive")
````

````python
lgb_path = os.path.join(BASE_PATH, 'models/lgb.joblib')

if not Path(lgb_path).exists():
    print("LightGBM Optuna Tuning (30 trials)")

    def objective_lgb(trial):
        params = {
            "n_estimators":      trial.suggest_int("n_estimators", 300, 1000, step=50),
            "max_depth":         trial.suggest_int("max_depth", 3, 8),
            "learning_rate":     trial.suggest_float("learning_rate", 0.005, 0.15, log=True),
            "num_leaves":        trial.suggest_int("num_leaves", 20, 150),
            "subsample":         trial.suggest_float("subsample", 0.6, 1.0),
            "colsample_bytree":  trial.suggest_float("colsample_bytree", 0.4, 0.9),
            "min_child_samples": trial.suggest_int("min_child_samples", 10, 50),
            "reg_alpha":         trial.suggest_float("reg_alpha", 1e-3, 5.0, log=True),
            "reg_lambda":        trial.suggest_float("reg_lambda", 1e-3, 5.0, log=True),
            "boosting_type":     trial.suggest_categorical("boosting_type", ["gbdt", "dart"]),
            "random_state": 42, "n_jobs": -1, "verbose": -1,
        }
        model = lgb.LGBMRegressor(**params)
        model.fit(
            X_train_raw, y_train,
            eval_set=[(X_val_raw, y_val)],
            callbacks=[lgb.early_stopping(50, verbose=False), lgb.log_evaluation(-1)],
        )
        return np.sqrt(mean_squared_error(y_val, model.predict(X_val_raw)))

    study_lgb = optuna.create_study(direction="minimize")
    study_lgb.optimize(objective_lgb, n_trials=30)
    print(f"Best LightGBM val RMSE (log): {study_lgb.best_value:.5f}")

    best_lgb = lgb.LGBMRegressor(
        **study_lgb.best_params, random_state=42, n_jobs=-1, verbose=-1
    )
    best_lgb.fit(X_train_val_raw, y_train_val)
    joblib.dump(best_lgb, lgb_path)
    print("LightGBM saved to Drive")
else:
    best_lgb = joblib.load(lgb_path)
    print("LightGBM loaded from Drive")
````

````python
stacking_path = os.path.join(BASE_PATH, 'models/stacking.joblib')

if not Path(stacking_path).exists():
    print('Building Stacking Ensemble (XGB + LightGBM)')

    xgb_params = best_xgb.get_params()
    lgb_params = best_lgb.get_params()

    xgb_pipe = Pipeline([('model', xgb.XGBRegressor(**xgb_params))])
    lgb_pipe = Pipeline([('model', lgb.LGBMRegressor(**lgb_params))])

    stacking_reg = StackingRegressor(
        estimators=[('xgb', xgb_pipe), ('lgb', lgb_pipe)],
        final_estimator=RidgeCV(alphas=(0.01, 0.1, 1.0, 10.0, 100.0)),
        cv=5, n_jobs=-1, passthrough=False,
    )
    stacking_reg.fit(X_train_val_raw, y_train_val)
    joblib.dump(stacking_reg, stacking_path)
    print('Stacking Ensemble saved to Drive')
else:
    stacking_reg = joblib.load(stacking_path)
    print('Stacking Ensemble loaded from Drive')
````

### 4.4 Inovasi Metodologi: Quantile Regression ($P_{10}, P_{50}, P_{90}$) untuk Koridor Negosiasi Upah

Pendekatan prediksi tunggal (*single point estimate*) tidak realistis dalam konteks ketenagakerjaan dunia nyata, karena upah suatu profesi selalu memiliki rentang fleksibilitas yang dipengaruhi oleh negosiasi, anggaran perusahaan, dan keunikan kandidat.

Untuk memberikan solusi aplikatif, dikembangkan model **Quantile Regression** menggunakan fungsi kerugian asimetris **Pinball Loss** (atau *Quantile Loss*):
$$\mathcal{L}_lpha(y, \hat{y}) = \max\Big(lpha(y - \hat{y}),\, (lpha - 1)(y - \hat{y})\Big) = egin{cases} lpha (y - \hat{y}) & 	ext{jika } y \ge \hat{y} \ (1 - lpha)(\hat{y} - y) & 	ext{jika } y < \hat{y} \end{cases}$$

#### Tiga Koridor Negosiasi yang Dihasilkan:
1. **Kuantil Bawah ($P_{10}, lpha = 0.10$):** Berperan sebagai batas bawah pengaman (*reservation wage floor*) bagi pencari kerja guna menghindari *underpaid*.
2. **Kuantil Tengah ($P_{50}, lpha = 0.50$):** Estimasi nilai tengah pasar (*fair market median*) yang menjadi patokan standar.
3. **Kuantil Atas ($P_{90}, lpha = 0.90$):** Batas atas kompetitif (*premium ceiling*) untuk kandidat berkinerja tinggi atau institusi dengan anggaran superior.

#### Metrik Validasi Kuantil:
* **Prediction Interval Coverage Probability (PICP):** Proporsi data aktual yang berada di dalam interval $[\hat{y}_{P_{10}}, \hat{y}_{P_{90}}]$:
  $$	ext{PICP} = rac{1}{N} \sum_{i=1}^{N} \mathbb{I}\left( \hat{y}_{i, P_{10}} \le y_i \le \hat{y}_{i, P_{90}} ight)$$
* **Mean Prediction Interval Width (MPIW):** Rata-rata lebar rentang negosiasi:
  $$	ext{MPIW} = rac{1}{N} \sum_{i=1}^{N} \left( \hat{y}_{i, P_{90}} - \hat{y}_{i, P_{10}} ight)$$


````python
print("Training Quantile Regressors (Alpha = 0.10 and Alpha = 0.90)...")

lgb_p10 = lgb.LGBMRegressor(
    objective='quantile', alpha=0.10, n_estimators=500, learning_rate=0.03,
    max_depth=6, num_leaves=31, subsample=0.8, colsample_bytree=0.7,
    random_state=42, n_jobs=-1, verbose=-1
)
lgb_p10.fit(X_train_val_raw, y_train_val)

lgb_p90 = lgb.LGBMRegressor(
    objective='quantile', alpha=0.90, n_estimators=500, learning_rate=0.03,
    max_depth=6, num_leaves=31, subsample=0.8, colsample_bytree=0.7,
    random_state=42, n_jobs=-1, verbose=-1
)
lgb_p90.fit(X_train_val_raw, y_train_val)

preds_p10_lin = np.expm1(lgb_p10.predict(X_test_raw))
preds_p90_lin = np.expm1(lgb_p90.predict(X_test_raw))
preds_p50_lin = preds_stack_lin

def pinball_loss(y_true, y_pred, alpha):
    diff = y_true - y_pred
    return np.mean(np.maximum(alpha * diff, (alpha - 1.0) * diff))

loss_p10 = pinball_loss(y_test, lgb_p10.predict(X_test_raw), 0.10)
loss_p90 = pinball_loss(y_test, lgb_p90.predict(X_test_raw), 0.90)

in_interval_mask = (y_test_linear >= preds_p10_lin) & (y_test_linear <= preds_p90_lin)
picp_coverage = in_interval_mask.mean() * 100.0
mean_interval_width = (preds_p90_lin - preds_p10_lin).mean()

quantile_metrics = pd.DataFrame([{
    "Metric": "Quantile Interval Analysis",
    "Coverage_PICP_Percent": round(picp_coverage, 2),
    "Target_Coverage_Percent": 80.0,
    "Mean_Interval_Width_USD": round(mean_interval_width, 2),
    "Pinball_Loss_P10": round(loss_p10, 5),
    "Pinball_Loss_P90": round(loss_p90, 5)
}])

quantile_metrics.to_csv(os.path.join(BASE_PATH, 'exports/tables_csv/table_quantile_metrics.csv'), index=False)
quantile_metrics.to_csv('exports/tables_csv/table_quantile_metrics.csv', index=False)

print(f"QUANTILE REGRESSION EVALUATION METRICS:")
print(f"Prediction Interval Coverage Probability (PICP): {picp_coverage:.2f}% (Target: ~80.0%)")
print(f"Mean Prediction Interval Width (MPIW)         : ${mean_interval_width:,.2f}")
print(f"Pinball Loss (Alpha = 0.10)                   : {loss_p10:.5f}")
print(f"Pinball Loss (Alpha = 0.90)                   : {loss_p90:.5f}")

sample_indices = np.argsort(y_test_linear)[:50]

plt.figure(figsize=(15, 6))
x_axis = np.arange(50)

plt.fill_between(x_axis, preds_p10_lin[sample_indices], preds_p90_lin[sample_indices],
                 color='#3498db', alpha=0.25, label='Quantile Prediction Interval (P10 - P90)')
plt.plot(x_axis, preds_p50_lin[sample_indices], color='#2980b9', lw=2, label='Predicted Median Salary (P50 - Stacking)')
plt.scatter(x_axis, y_test_linear[sample_indices], color='#e74c3c', s=25, zorder=5, label='Actual Test Salary')

plt.title(f'SkillGraph-AI: Salary Negotiation Prediction Intervals (PICP Coverage = {picp_coverage:.1f}%)', fontweight='bold', fontsize=13)
plt.xlabel('Sorted Sample Test Job Postings', fontweight='bold')
plt.ylabel('Annual Salary (USD/yr)', fontweight='bold')
plt.gca().yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x*1e-3:.0f}K'))
plt.legend(loc='upper left', fontsize=10)
plt.tight_layout()
plt.savefig(os.path.join(BASE_PATH, 'figures/fig_quantile_intervals.png'), dpi=300, bbox_inches='tight')
plt.savefig('exports/figures/fig_quantile_intervals.png', dpi=300, bbox_inches='tight')
plt.show()

fig_plotly_quant = go.Figure()
fig_plotly_quant.add_trace(go.Scatter(
    x=x_axis, y=preds_p90_lin[sample_indices], mode='lines',
    line=dict(width=0), showlegend=False, name='P90 Optimistic'
))
fig_plotly_quant.add_trace(go.Scatter(
    x=x_axis, y=preds_p10_lin[sample_indices], mode='lines',
    line=dict(width=0), fill='tonexty', fillcolor='rgba(52, 152, 219, 0.25)',
    name='80% Negotiation Interval (P10 - P90)'
))
fig_plotly_quant.add_trace(go.Scatter(
    x=x_axis, y=preds_p50_lin[sample_indices], mode='lines',
    line=dict(color='#2980b9', width=2), name='P50 Median Benchmark'
))
fig_plotly_quant.add_trace(go.Scatter(
    x=x_axis, y=y_test_linear[sample_indices], mode='markers',
    marker=dict(color='#e74c3c', size=6), name='Actual Salary'
))
fig_plotly_quant.update_layout(
    title="<b>Interactive Salary Negotiation Range Simulator (Quantile Interval P10 - P90)</b>",
    xaxis_title="Job Sample", yaxis_title="Annual Salary (USD)",
    yaxis_tickprefix='$', yaxis_tickformat=',.0f', template='plotly_white'
)
fig_plotly_quant.write_html(os.path.join(BASE_PATH, 'exports/plotly_html/plotly_quantile_salary.html'))
fig_plotly_quant.write_html('exports/plotly_html/plotly_quantile_salary.html')
print("Saved fig_quantile_intervals.png (300 DPI) and interactive plotly_quantile_salary.html")
````

### 4.5 Inovasi Metodologi: Explainable AI dan Valuasi Ekonomi Marjinal Berbasis TreeSHAP

Model pohon ansambel kerap dikritik sebagai model kotak hitam (*black-box*). Dalam penetapan upah, ketiadaan transparansi dapat memicu ketidakpercayaan dan risiko bias diskriminatif.

Untuk memberikan interpretabilitas penuh, diterapkan algoritma **SHAP (SHapley Additive exPlanations) TreeExplainer** yang berakar pada teori permainan kooperatif (*cooperative game theory*). Nilai kontribusi marjinal Shapley $\phi_j$ dihitung secara eksak untuk setiap fitur $j$:
$$f(x) = \phi_0 + \sum_{j=1}^{M} \phi_j(x)$$

#### Formulasi Konversi Moneter Nyata (USD/tahun):
Karena model regresi memprediksi target dalam skala logaritma natural $\ln(1 + 	ext{salary})$, nilai atribusi $\phi_j$ ditransformasikan ke dalam nilai moneter riil dalam Dolar AS per tahun menggunakan nilai median kompensasi korpus ($	ext{Salary}_{	ext{Median}} = \$90.000$):
$$\Delta 	ext{Salary}_{	ext{USD}} = 	ext{Salary}_{	ext{Median}} 	imes \left( e^{|\phi_j|} - 1 ight)$$
Formulasi non-linier ini memungkinkan penghitungan *return-on-investment* (ROI) kuantitatif dari setiap keahlian teknis dan sertifikasi profesional secara langsung.


````python
print("Computing SHAP values on representative test sample...")

feat_names_proposed = (
    [f"sbert_{i}" for i in range(sbert_embeddings.shape[1])]
    + [f"tfidf_{i}" for i in range(tfidf_embeddings.shape[1])]
    + num_cols_proposed
)

X_shap_sample = X_test_raw[:500]
explainer     = shap.TreeExplainer(best_lgb)
shap_values   = explainer.shap_values(X_shap_sample)

plt.figure(figsize=(12, 8))
shap.summary_plot(shap_values, X_shap_sample, feature_names=feat_names_proposed, max_display=20, show=False)
plt.title("SkillGraph-AI: SHAP Summary Beeswarm Plot (Top 20 Features on Log Salary)", fontsize=13, fontweight='bold', pad=20)
plt.tight_layout()
plt.savefig(os.path.join(BASE_PATH, 'figures/fig_shap_beeswarm.png'), dpi=300, bbox_inches='tight')
plt.savefig('exports/figures/fig_shap_beeswarm.png', dpi=300, bbox_inches='tight')
plt.show()

baseline_median_salary = float(df_reg['salary_mid'].median())
mean_abs_shap = np.abs(shap_values).mean(axis=0)

shap_monetary_df = pd.DataFrame({
    'feature_name': feat_names_proposed,
    'mean_abs_shap_log': mean_abs_shap,
    'marginal_impact_usd': baseline_median_salary * (np.exp(mean_abs_shap) - 1.0)
}).sort_values('marginal_impact_usd', ascending=False)

top15_monetary = shap_monetary_df.head(15).copy()
top15_monetary.to_csv(os.path.join(BASE_PATH, 'exports/tables_csv/table_shap_monetary_valuation.csv'), index=False)
top15_monetary.to_csv('exports/tables_csv/table_shap_monetary_valuation.csv', index=False)

print("TOP 15 MARKET MONETARY DRIVERS (Marginal Value in USD/year)")
print(top15_monetary[['feature_name', 'mean_abs_shap_log', 'marginal_impact_usd']].to_string(index=False))

plt.figure(figsize=(12, 6.5))
bars = plt.barh(top15_monetary['feature_name'][::-1], top15_monetary['marginal_impact_usd'][::-1], color='#16a085', edgecolor='black', alpha=0.85)
plt.title('SkillGraph-AI: Top 15 Feature Monetary Drivers (Marginal Economic Valuation)', fontweight='bold', fontsize=13)
plt.xlabel('Marginal Annual Impact (USD/year)', fontweight='bold')
plt.gca().xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x*1e-3:.0f}K'))
for bar in bars:
    w = bar.get_width()
    plt.text(w + 200, bar.get_y() + bar.get_height()/2, f'+${w:,.0f}/yr', va='center', fontweight='bold', fontsize=9)
plt.xlim(0, top15_monetary['marginal_impact_usd'].max() * 1.15)
plt.tight_layout()
plt.savefig(os.path.join(BASE_PATH, 'figures/fig_shap_monetary_bar.png'), dpi=300, bbox_inches='tight')
plt.savefig('exports/figures/fig_shap_monetary_bar.png', dpi=300, bbox_inches='tight')
plt.show()
print("Saved fig_shap_beeswarm.png, fig_shap_monetary_bar.png, and table_shap_monetary_valuation.csv")
````
